## Chapter 6 Judgmental forecasts

Judgmental forecasting is highly subjective but essential in situations where historical data is completely lacking (e.g., launching a new product, unprecedented market conditions), incomplete, or only available with a delay. It is also commonly used to adjust statistical forecasts when forecasters possess important domain knowledge about upcoming events.

### 6.1 Beware of limitations
Because judgmental forecasts depend heavily on human cognition, they are vulnerable to several biases and limitations:
*   **Inconsistency:** Human judgment can vary day-to-day based on mood, limited attention spans, or a tendency to weigh recent events more heavily than older ones.
*   **Personal and Political Agendas:** Judgment is often clouded if the forecaster is also responsible for setting targets; they may forecast low to easily exceed expectations.
*   **Optimism/Wishful Thinking:** Especially in group settings or when launching new products, forecasters often suffer from over-enthusiasm.
*   **Anchoring:** Forecasters tend to be unduly influenced by a familiar reference point (like the last observed value) and fail to properly adjust for new information.

### 6.2 Key principles
To reduce the adverse effects of human bias, judgmental forecasting must follow a well-structured, systematic approach based on five key principles:
1.  **Set the forecasting task clearly and concisely:** Avoid ambiguous terms and irrelevant information that might distract the forecaster.
2.  **Implement a systematic approach:** Use checklists and decision rules to identify what information is relevant and how it should be weighted.
3.  **Document and justify:** Forcing forecasters to write down their rules and assumptions promotes consistency and accountability, reducing bias.
4.  **Systematically evaluate forecasts:** Keep records of forecasts and compare them against actual observations to provide feedback and improve future accuracy.
5.  **Segregate forecasters and users:** Those producing the forecast should be completely separate from the users implementing plans or setting business targets. A forecast should reflect reality, not managerial hopes.

### 6.3 The Delphi method
The Delphi method is built on the assumption that consensus forecasts from a group are more accurate than individual forecasts. It involves a structured, iterative process managed by a facilitator:
1.  **Panel of Experts:** Assemble 5 to 20 experts with diverse skills.
2.  **Anonymity:** Experts submit their initial forecasts and justifications anonymously. This prevents senior members or loud personalities from dominating the group, and suppresses political pressures.
3.  **Iterative Feedback:** The facilitator compiles the forecasts (using numerical summaries and highlighting outliers) and sends them back to the experts. 
4.  **Review:** Experts adjust their forecasts based on the feedback. This iterates (usually 2-3 rounds) until the variability of the responses decreases to a satisfactory consensus.
5.  **Final Forecast:** Typically constructed by equally weighting all the experts' final estimates.

### 6.4 Forecasting by analogy
This approach involves generating forecasts by comparing the target situation to similar historical properties or events (e.g., an appraiser pricing a house based on similar nearby sales). 
To minimize bias, it is best applied using a **structured analogy** approach: a panel of experts independently identifies multiple analogous situations, rates their similarity to the target task, and the facilitator derives a final forecast using a weighted rule based on the similarity rankings.

### 6.5 Scenario forecasting
Instead of trying to predict the single most likely outcome, scenario forecasting generates forecasts based on a wide range of plausible scenarios (typically "best", "middle", and "worst" cases). By exploring all possible drivers and their interactions, organizations can identify extreme outcomes and engage in early contingency planning.

### 6.6 New product forecasting
When a product is entirely new or entering a new market, historical data does not exist. Alongside the Delphi and analogy methods, three specific (though less structured) methods are used:
*   **Sales force composite:** Aggregating forecasts from salespeople. *Warning:* Salespeople are close to the customer, but they lack forecasting training and may introduce severe self-serving biases (setting forecasts low to easily meet targets, or setting them high out of enthusiasm).
*   **Executive opinion:** A group meeting of top-level management. *Warning:* While this brings diverse skills together, it is highly susceptible to group meeting biases and optimism.
*   **Customer intentions:** Using structured questionnaires to ask customers their likelihood of purchasing the product. *Warning:* There is often a strong disconnect between what customers say they will buy and their actual purchasing behavior, especially if the product is completely novel rather than just an update.

### 6.7 Judgmental adjustments
This applies when historical data *is* available, and statistical models have already generated a forecast. Practitioners often manually adjust these statistical forecasts, but they must do so carefully:
*   **Use adjustments sparingly:** Only adjust a statistical forecast if you have *significant additional information* that the model cannot know (e.g., upcoming promotions, holidays, strikes). 
*   **Avoid small adjustments:** Do not attempt to manually correct a statistical model because you think you spot a "pattern" in the data. Statistical models are far better at handling noisy data; small judgmental adjustments generally hinder accuracy and only serve to give the forecaster an "illusion of optimism".
*   **Apply a structured approach:** If you must adjust, use a Delphi-like panel and require all adjustments to be strictly documented and justified to guard against unnecessary overrides.
